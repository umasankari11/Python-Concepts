## **User-Defined Functions**


```python

# ---------------------------
# 1. USER-DEFINED FUNCTIONS
# ---------------------------

# Basic function with no parameters

```

```python
def profile():
  print("Name : Umasankari, Qualification : M.Tech")
profile()
```

```python

# Function with parameters

def add(num1,num2):
  total=num1+num2
  print(total)
add(1,2)



```

```python
def addition(num1,num2):
  total=num1+num2
  return total

tot=addition(34,12)
tot
```

```python
# ---------------------------
#  RETURN STATEMENT
# ---------------------------
# 'return' sends a value back and ends the function execution

def check_num(num):
  if num>0:
    return "Positive"
  elif num<0:
    return "Negative"
  else:
    return "0"

check_num(-1)
```

```python

# Function with multiple returns
def check_num(num):
  if num>0:
    return "Positive"
  elif num<0:
    return "Negative"
  else:
    return "0"

check_num(-1)
```

```python
# ---------------------------
#  YIELD STATEMENT (Generators)
# ---------------------------
# 'yield' creates a generator → returns one value at a time
# Does NOT end the function (unlike return)


# Another generator example
def squares(nums):
    for n in nums:
        yield n*n   # yields each square

for sq in squares([1, 2, 3, 4]):
    print("Square:", sq)
```

write a function which get your name qualification and salary==>calculate your salary =salary+10000--> returns your updated salary

```python
def salary(name, qualification, salary):
    salary = salary + 10000
    yield f"Name: {name}"
    yield f"Qualification: {qualification}"
    yield f"Updated Salary: {salary}"



for updated_info in salary("Uma", "M.Tech", 30000):
    print(updated_info)
```

## **Lambda Functions**

```python
# ---------------------------
#LAMBDA FUNCTIONS
# ---------------------------
# Anonymous (nameless) function
# Syntax: lambda arguments : expression

square = lambda x: x * x
print("Square of 5:",square(5))

```

```python

# Lambda with multiple arguments
add_three = lambda a, b, c: a + b + c
print(add_three(10, 20, 30))

```

```python
# Lambda inside filter()
nums = [1, 2, 3, 4, 5]

greater_than_2 = list(filter(lambda x: x > 2, nums))
print("Greater than 2:", greater_than_2)
```

```python
#create lambda function which return likst of characters in your name

name_to_chars = lambda name: list(name)

print(name_to_chars("Umasankari"))
```

# **TYPES OF ARGUMENTS + LOCAL & GLOBAL VARIABLES**

```python
# ------------------------------------------------------------
# TYPES OF ARGUMENTS + LOCAL & GLOBAL VARIABLES
# ------------------------------------------------------------


# ============================================================
# 1. TYPES OF ARGUMENTS IN PYTHON FUNCTIONS
# ============================================================

# ------------------------------------------------------------
# A. POSITIONAL ARGUMENTS
# ------------------------------------------------------------
def details(name,age):
  print(name,age)


 # values must be in correct order

details("Uma",35)
```

```python

# ------------------------------------------------------------
# B. KEYWORD ARGUMENTS
# ------------------------------------------------------------
details(age=22, name="Asha")   # order doesn't matter when using keywords


# ------------------------------------------------------------
# C. DEFAULT ARGUMENTS
# ------------------------------------------------------------
def greet(name="Guest"):
    print("Hello,", name)

print(greet())            # uses default value
print(greet("Meena"))     # overrides default value


```

```python

# ------------------------------------------------------------
# D. VARIABLE-LENGTH ARGUMENTS ( *args )
# ------------------------------------------------------------
# *args = accepts multiple positional arguments as a tuple

def total_marks(*marks):
    print("Marks Tuple:", sum(marks))


total_marks(90, 85, 78, 92)

```

```python

# ------------------------------------------------------------
# E. VARIABLE-LENGTH KEYWORD ARGUMENTS ( **kwargs )
# ------------------------------------------------------------
# **kwargs = accepts multiple keyword arguments as a dictionary

def show_profile(**info):
    print("Profile Dictionary:", info)

show_profile(name="Anil", age=25, course="Python", city="Delhi")

```

```python


# ============================================================
# 2. LOCAL vs GLOBAL VARIABLES
# ============================================================

# ------------------------------------------------------------
# LOCAL VARIABLE
# ------------------------------------------------------------
def test_local():
    x = 5      # local variable (exists only inside function)
    print("Inside function, x =", x)

test_local()
print(x)
# print(x)     # This will cause error: x is not defined
```

```python

# ------------------------------------------------------------
# MODIFYING GLOBAL VARIABLE USING 'global'
# ------------------------------------------------------------
count = 0  # global variable

def increase():
    global count      # telling Python to use global 'count'
    count += 1

increase()
increase()
increase()
increase()
increase()
print("Count after calling function:", count)

```
