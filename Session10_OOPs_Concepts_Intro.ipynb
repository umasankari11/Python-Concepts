### **Make sensitive fields private**

```python
class BankAccount():
  def __init__(self,balance):
    self.__balance=balance
  def get_balance(self):
    return self.__balance
  def deposit(self,amount):
    self.__balance+=amount

acc=BankAccount(3000)
print("Balance Amount",acc.get_balance())
acc.deposit(2000)
print("Balance Amount",acc.get_balance())




```

### **Getters & Setters**

```python
class Person():
  def __init__(self,name):
    self.p_name=name
  def get_name(self):
    return self.p_name
  def set_name(self,name):
    self.p_name=name
p=Person("Mike")
print("Before setter method:",p.get_name())
p.set_name("Anju")
print("After setter method:",p.get_name())
```

# ⭐ **Polymorphism**

## **📘 Explanation**

Polymorphism = one interface, multiple implementations.


## **🎯 Single Inheritance**



```python
class Animal():
  def sound(self):
    print("Some sound")
class dog(Animal):
  def sound(self):
    super().sound()
    return "Bark"
d=dog()
d.sound()
```

### **Multilevel INHERITANCE**

```python
class A():
  def showA(self):
    return "This is A"
class B(A):
  def showB(self):
    return "This is B"
class C(B):
  def showC(self):
    return "This is C"
obj=C()
print(obj.showA())
print(obj.showB())
print(obj.showC())
```

### **Multiple Inheritance**

```python
class Parent1():
  def showP1(self):
    return "This is Parent 1"
class Parent2():
  def showP2(self):
    return "This is PArent 2"
class Child(Parent1,Parent2):
  pass

objc=Child()
print(objc.showP1())

print(objc.showP2())
```

### **Polymorphism (Same Method, Different Action)**

```python
class shape():
  def area(self):
    return "No area defined"

class Square(shape):
  def __init__(self,side):
    self.side=side
  def area(self):
    return self.side*self.side

class Circle(shape):
  def __init__(self,radius):
    self.radius=radius
  def area(self):
    return 3.14*self.radius*self.radius

s=Square(3)
c=Circle(2)
print(c.area())
print(s.area())
print(c.area())
```

## **Static Methods**

 A method that does not depend on:object,class

**Used for:**

- Utility/helper calculations

**Example:** Validate license number.

**Uses:** @staticmethod



```python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        return "Car Started"

obj = Car()
print(obj.start())
```

### **Class Method**

```python
class Student:
    school = "ABC School"

    @classmethod
    def get_school(cls):
        return cls.school

print(Student.get_school())
```

### **Static Method**

```python
class Math:
    @staticmethod
    def is_even(num):
        return num % 2 == 0

print("Is 10 Even?", Math.is_even(10))
```

## **Composition (HAS–A Relationship)**

```python
class Engine:
    def start(self):
        return "Engine Started"

class Car:
    def __init__(self):
        self.engine = Engine()  # HAS-A relationship

c = Car()
print(c.engine.start())
```
