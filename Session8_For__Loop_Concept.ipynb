<b>The `range()` Function</b><br>

*   To loop through a set of code a specified number of times, we can use the `range()` Function.
*   The `range()` function returns a sequence of numbers, starting from 0 by default, and increments by 1 (by default), and ends at a specified number.



<br>

**syntax of range()**

`range(start_num,end_num,step)`
<br><br>

*   The `range()` function defaults to 0 as a starting value, however it is possible to specify the starting value by adding a parameter:
>>`range(2, 10)`

 which means values from 2 to 10 (but not including 10).

<font color = "#ef7300" size="5px"><b><u> Jumping statements in python :</font></b><br></u>

 **Jumping statements are used in both While and For loops.**

<font color = "#0073eb" size="3px"><b><u> Break statement :</font></b><br></u>
The `break statement` we can stop the loop before it has looped through all the items.
<font color="ef7300" size="4px" >   
     <b> <u>Activity :-</u></b></font>





````python
# Activity - create a for loop Using jumping statements
# Use Break statement:
num=range(2,20,2)#2,4,6,8,10,12,14,16,18
for i in num:
  if i==12:
    break
  print(i)
````

<font color = "#0073eb" size="3px"><b><u> Continue statement :</font></b><br></u>
 The `continue statement` we can stop the current iteration of the loop, and continue with the next.
 <font color="ef7300" size="4px" >   
     <b> <u>Activity :-</u></b></font>

````python
#  Activity - create a for loop Using jumping statements
# Use Continue statement:
num=range(2,20,2)#2,4,6,8,10,12,14,16,18
for i in num:
  if i==12:
    continue
  print(i)
````

- `enumerate()` is a built-in Python function.
- It allows us to **loop over an iterable (like a list or string)** and automatically keeps track of the index (position) of each item.
- Syntax:  
  
  ```python
  enumerate(iterable, start=0)








````python
# Example 1: Without enumerate
fruits=["apple","orange","grapes"]
index=0
for i in fruits:
  print(index,i)
  index+=1
````

````python
### Example 2: With enumerate
for ind,fruit in enumerate(fruits):
  print(ind,fruit)


````

✅ Output will be the same, but `enumerate` makes it cleaner.

* Syntax:

  ```python

````python

### Example 1: Normal for loop
num_list=[]
for i in range(20):
  num_list.append(i)
num_list
````

````python
#Example 2: Using list comprehension
numb_list=[i for i in range(10)]
numb_list
````

````python
squares = [i**2 for i in range(10)]
print(squares)
````


## **What is `zip()` in Python?**

`zip()` is a built-in Python function that **combines multiple iterables** (like lists, tuples, strings) **element-wise** into pairs (or tuples).
It’s like making bundles from multiple sequences.













````python
names = ["Alice", "Mary", "Punitha"]
age = [34, 45, 67]

for n, a in zip(names,age):
    print(n, a)
````

##  How it Works

* It takes the **1st element** from each iterable and pairs them.
* Then the **2nd elements**, and so on.
* It stops at the **shortest iterable**.

````python
names = ["Asha", "Ravi", "Geeta"]
marks = [90, 85, 88]

zipped_list = list(zip(names, marks))
print(zipped_list)


````

##  **Convert Zipped Values to Other Structures**

### To a list:


````python
[0 for _ in range(3)]
````

### To a dictionary:


````python
names = ["Asha", "Ravi", "Geeta"]
marks = [90, 85, 88]

zipped_dict = dict(zip(names, marks))
print(zipped_dict)
````


##  Unzipping (Reverse Zip)

If you zipped data and want to separate it again:

````python
name,age=zip(*zipped_list)
print(name)
print(age)
````

##  **Real-world Usage Example**

### **Looping through two lists at the same time:**

```python
for question, answer in zip(q_list, a_list):
    print(f"Q: {question} | A: {answer}")
```

### **Summing values from two lists:**

```python
[a + b for a, b in zip([1, 2, 3], [4, 5, 6])]
```

**Output:** `[5, 7, 9]`

 <font color = "#0073eb" Size = 4px> <b><u>Create Pattern Printing (Square) Program : </u></b></font>
  <br>



**Algorithm:**

  
   
- Ask the user to input the size of the square.
   - Read and store the user input size in a variable called `size`.

*  Start a loop to iterate over the range of values from 0 to `size-1` (inclusive). Let the loop variable be `i`.

*  Inside the loop:
   - Print `size` asterisks (`*`) horizontally.
   - To achieve this, use the `print()` function with `"*"` multiplied by `size`.

*  Repeat step 5 for each value of `i` in the range.

*  End loop.
* End of the `main()` function.

* Call the `main()` function to execute the program.









````python
def main():
    # Step 1: Ask user for size
    size = int(input("Enter the size of the square: "))

    # Step 2: Loop from 0 to size-1
    for i in range(size):

        # Step 3: Print size number of *
        print("*" * size)

# Step 4: Call the function
main()
````
