    <b><u>  Index() :</u></b></font>

*  The `index()` method finds the first occurrence of the specified value.

*  The `index()` method raises an exception if the value is not found.

<font color="#0073eb" size="4px" >
    <b><u>  syntax :</u></b></font>

   `string.index(value, start, end)`

````python
# Teacher and student activity  :-
msg = "Hello ! welcome to Entri."

var=msg.index("e",2,12)
var
````

<font color = "#0073eb" size="4px" ><b><u>  Code Explanation :- </b> </u>
</font> <br>

*  `msg = "Hello ! welcome to Entri."`: This line initializes a string variable `msg` containing the text "Hello ! welcome to Entri."

*  `var = msg.index("e", 2, 12)`: This line calls the `index()` method on the string `msg`. The `index()` method searches for the specified substring `"e"` within the string. The second argument `2` specifies the starting index of the search (inclusive), and the third argument `12` specifies the ending index of the search (exclusive). This means that the search will be conducted starting from the character at index `2` up to (but not including) the character at index `12`.

*  The `index()` method returns the index of the first occurrence of the substring `"e"` within the specified range of indices. If the substring is not found within the specified range, a `ValueError` is raised.

*  `print(var)`: This line prints the value of the variable `var`, which contains the index of the first occurrence of the substring `"e"` within the specified range.

   * **Using the List/array slicing  [ :: ]  method**


````python
# Teacher and student activity  :-


# String slicing using slice() method
print(string_1)

s1=slice(3)
s2=slice(3,8,2)

print(string_1[s1])
print(string_1[s2])


# String slicing using list/array[] method

#str[start:stop]  2 to 6
print(string_1[2:7])

#str[start:]
print(string_1[4:])

#str[:stop]
print(string_1[:8])
#str[:] # a copy of the whole array
print(string_1[:])
#str[start:stop:step]

print(string_1[3:8:2])
````

 **Palindrome:**

  A palindrome string is like a word or a sentence that reads the same forwards and backwards. For example, "racecar" is a palindrome because if you read it from left to right or right to left, it's still "racecar". It's like a cool word puzzle!


### **Activity: Let's create a program which check whether the give string is a palindrome or not**

````python
#Challenge
# Program: Check if a string is a palindrome

# Step 1: Take input from the user
text = input("Enter a string: ")

# Step 2: Remove spaces and convert to lowercase for uniform comparison
clean_text = text.replace(" ", "").lower()

# Step 3: Check if the string is equal to its reverse
if clean_text == clean_text[::-1]:
    print(f'"{text}" is a palindrome!')
else:
    print(f'"{text}" is not a palindrome.')
````

<font color = "#0073eb" size="4px" ><b><u>  Code Explanation :- </b> </u>
</font> <br>


1. `def is_palindrome(string):`: This line defines a function called `is_palindrome`. A function is like a small program that performs a specific task. This function checks if a given `string` is a palindrome or not. A palindrome is a word, phrase, number, or other sequence of characters that reads the same forward and backward.

2. `return string == string[::-1]`: This line contains a comparison statement. It checks if the given `string` is equal to its reverse. `string[::-1]` is a way to reverse the string. If the string is equal to its reverse, it means the string is a palindrome. The `return` keyword means the function will give back the result of this comparison.

3. `string = input()`: This line takes input from the user and stores it in the variable named `string`. `input()` is a function that allows the user to type something, and whatever they type is stored as a string.

4. `if is_palindrome(string):`: This line checks if the function `is_palindrome` returns `True` for the `string` inputted by the user. If the function returns `True`, it means the string is a palindrome, so it prints `"The string is a palindrome."`. Otherwise, it prints `"The string is not a palindrome."`. This `if` statement helps the program decide what message to display based on whether the inputted string is a palindrome or not.

     <b><u>Cleaning and Formatting Customer Feedback</u></b></font>
 <br>

**Scenario:**
We received raw feedback data from an online survey.  
The text has inconsistent capitalization, extra spaces, and punctuation errors.  
We need to clean it for use in a dashboard.


````python
# Case Study Data
feedback = input("Feedback please!")

# Step 1: Remove leading/trailing spaces
feedback_tra_lead=feedback.strip()

# Step 2: Standardize case
feedback_lower=feedback_tra_lead.lower()

# Step 3: Remove extra punctuation
pun_removed=feedback_lower.replace("!!","!")

# Step 4: Format final output
print(f'User Feedback:{pun_removed}')
````

### Output:
`Customer Feedback: Great course , loved the content!`

**Scenario:**
You received raw HR data containing inconsistent formatting.  
You need to extract the employee name, email, and ID properly.

**Data:**

record = " EmployeeName :  john doe , Email : JOHN.DOE@company.COM , ID :  emp_0087 "


Task Instructions:
Use string methods and slicing to:
1. Remove unwanted spaces.
2. Convert:
   - Name → `John Doe`
   - Email → `john.doe@company.com`
3. Extract the Employee ID and print only the numeric part: `0087`


````python
record = " EmployeeName : john doe , Email : JOHN.DOE@company.COM , ID : emp_0087 "
````

````python
# Step 1: Clean the string
cleaned_record=record.strip()
cleaned_record
````

````python
cleaned_record.index("j",0)
````

````python
# Step 2: Extract name

employee_name=cleaned_record[15:23]
employee_name
````

````python

# Step 3: Extract email

email_start = cleaned_record.index("JOHN.DOE")
email_end = cleaned_record.index(",", email_start)
employee_email = cleaned_record[email_start:email_end].lower()
print("Employee Email:", employee_email)
````

````python

# Step 4: Extract employee ID and get numeric part

id_start = cleaned_record.index("emp_")
employee_id = cleaned_record[id_start + 4:].strip()
print("Employee ID:", employee_id)
````
