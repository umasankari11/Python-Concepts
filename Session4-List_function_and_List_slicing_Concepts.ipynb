## **2. List Methods Overview**



| Category               | Method                             | Description         |
| ---------------------- | ---------------------------------- | ------------------- |
| **Adding Elements**    | `append()`, `insert()`, `extend()` | Add items           |
| **Removing Elements**  | `remove()`, `pop()`, `clear()`     | Remove items        |
| **Rearranging**        | `sort()`, `reverse()`              | Sort or reverse     |
| **Searching/Counting** | `index()`, `count()`               | Find or count items |
| **Copying**            | `copy()`                           | Duplicate a list    |

## **3. Adding Elements**

```python
fruits=["Apple","Mango","Grape"]
print(fruits)
# append() → Adds to end
#l1=["Dragonfruit","Orange","Banana"]
fruits.append("Papaya")
print(f'Append list: {fruits}')

# insert() → Adds at specific index
fruits.insert(1,"Cherry")
print(f"Item Inserted in fruits:{fruits}")

# extend() → Adds multiple items

fruits.extend(l1)
print(f'Extendes list:{fruits}')
```

## **5. Rearranging Lists**

```python
numbers = [5, 2, 9, 1, 7]

# sort() → Ascending
numbers.sort()
print(f'Asc ordered sorted list\n{numbers}')

# sort(reverse=True) → Descending
numbers.sort(reverse=True)
print(f'Desc ordered sorted list\n{numbers}')

# reverse() → Reverses current order
numbers.reverse()
numbers
```

## **6. Searching and Counting**

```python
data = ["A", "B", "C", "A", "D", "A"]

# count() → Count occurrences
#count_A=data.count("A")
print(f'No of times: {data.count("A")}')
# index() → Find position
print(f'No of times: {data.index("C")}')
```

```python
data.pop(data.index("C"))#--data.index("C")--3==>data.pop(3)
```

# **9. List Slicing**


List slicing allows you to extract portions of a list using the syntax:

`**Syntax**`
> `list[start:end:step]`

```python
mat_list=[1,3,2,4,8,22,16,7]
# print 1st 4 item
mat_list[:4]
```

```python
#print item in the index from -2 to -5
mat_list[-2:-5:-1]
```

```python
#reverse the list
mat_list[::-1]
```
