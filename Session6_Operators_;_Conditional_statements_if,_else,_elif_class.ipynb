##  **1. Operators in Python**

Operators are symbols that perform operations on variables and values.

<br>
<br>


---

----

###  **Types of Operators**

| Type                     | Description                     | Example                                  |
| :----------------------- | :------------------------------ | :--------------------------------------- |
| **Arithmetic Operators** | Perform mathematical operations | `+`, `-`, `*`, `/`, `//`, `%`, `**`      |
| **Comparison Operators** | Compare two values              | `==`, `!=`, `>`, `<`, `>=`, `<=`         |
| **Assignment Operators** | Assign values to variables      | `=`, `+=`, `-=`, `*=`, `/=`, `//=`, `%=` |
| **Logical Operators**    | Combine multiple conditions     | `and`, `or`, `not`                       |
| **Membership Operators** | Check presence in sequence      | `in`, `not in`                           |
| **Identity Operators**   | Compare object identities       | `is`, `is not`                           |

---




```python
# Arithmetic Operators
a,b=15,4

print(f"a + b={a+b}")
print(f"a - b={a-b}")
print(f"a * b={a*b}")
print(f"a / b={a/b}")
print(f"a //b={a//b}")
print(f"13 % 2={13%2}")
print(f"3**3={3**3}")
```

Comparison Operators | Compare two values | ==, !=, >, <, >=, <= |

```python

# Comparison Operators
print(f"13 == 2=> {13==2}")
print(f"13 == 13=> {13==13}")
print(f"13 > 2=> {13>2}")
print(f"13 >=13=> {13>=13}")
print(f"13 < 2=> {13<2}")
print(f"4 <7=> {4<7}")
print(f"23 <=23 => {23<=23}")
print(f"13 != 2=> {13!=2}")
```

| Assignment Operators | Assign values to variables | =, +=, -=, *=, /=, //=, %= |

```python
a=4
# a=a+6
a+=6
print('a+=6',a)

a-=2
print('a-=2',a)
a*=10
print('a*=10',a)

a/=2
print('a/=2',a)

a//=3
print('a//=3',a)

a%=3
print('a%=3',a)
```

```python

# Logical Operators
x,y=True,False

print((22>14) and (21>20))
print(x or y)
print((22>14) or (21>27))

print("NOT",not x)
print(not((22>14) and (21>20)))
```

Membership Operators	Check presence in sequence	in, not in

```python

# Membership Operators

string1="Python"
print("y" in string1)
print("z" in string1)

print("------------Not In------------")
print("y" not in string1)
print("z" not in string1)
```

```python
# Identity Operators
p=[1,3,4]
q=[1,3,4]

dup=p
print("p is same as q",p is q)
print("p is same as dup",p is dup)

print("p is same as q",p is not q)
print("p is same as dup",p is not dup)


```

## 🔹 **3. Conditional Statements**

Conditional statements are used to make decisions based on specific conditions.

---

### 🔸 **if Statement**

```python
mark=85
total=200

if mark>=80:
  print("Good Job")

if mark<=85 and total==200:
  print("First class")
```

### **if–else Statement**

```python
age=20

if age > 18:
  print("He/She is eligible for voting")

else:
  print("He/She is not eligible for voting")


```


### **if–elif–else Statement**

```python
mark=20

#mark>=80-A+, mark>=75 and mark<80 =>A, mark>=60 and mark<75==>B, mark>=40 and mark<60==>c, mark<40==>D

if mark >= 80:
    grade = "A+"
elif mark >= 70 and mark < 80:
    grade = "A"
elif mark >= 60 and mark < 75:
    grade = "B"
elif mark >= 40 and mark < 60:
    grade = "C"
else:
    print("Fail")
    grade = "D"

print(f"Mark={mark}\n\n Grade:", grade)
```


##  **5. Case Study: Bank Loan Eligibility System**
### **Objective:**

Use **operators** and **conditional statements** to determine a customer’s **loan approval status** based on income, credit score, and age.

---

```python
# Step 1: Customer Data
customer_name = "Ravi"
age = 18
income = 55000
credit_score = 500
existing_loan = False

# Step 2: Loan Eligibility Check
if age >= 21 and income >= 30000:
    if credit_score >= 700 and not existing_loan:
        status = "Approved "
    elif credit_score >= 600 and income >= 40000:
        status = "Conditionally Approved "
    else:
        status = "Rejected "
else:
    status = "Not Eligible "

# Step 3: Display Result
print("Customer:", customer_name)
print("Age:", age)
print("Income:", income)
print("Credit Score:", credit_score)
print("Loan Status:", status)
```

## **Case Study Scholarship Eligibility Checker**

**Instructions:**

1. Create variables: `student_name`, `marks`, `attendance`, `family_income`.
2. Use conditional statements (`if`, `elif`, `else`) to:

   * Award **Full Scholarship** if marks ≥ 90 and attendance ≥ 90.
   * Award **Half Scholarship** if marks ≥ 80 or attendance ≥ 85.
   * Award **Need-Based Scholarship** if marks ≥ 70 and `family_income` < 30000.
   * Otherwise, display **Not Eligible**.
3. Display all data (no loops).


```python
student_name = "Uma"
marks = 98
attendance = 95
family_income = 25000


if marks >= 90 and attendance >= 90:
    scholarship = "Full Scholarship"
elif marks >= 80 or attendance >= 85:
    scholarship = "Half Scholarship"
elif marks >= 70 and family_income < 30000:
    scholarship = "Need-Based Scholarship"
else:
    scholarship = "Not Eligible"


print("Student Name:", student_name)
print("Marks:", marks)
print("Attendance:", attendance)
print("Family Income:", family_income)
print("Scholarship Status:", scholarship)
```
