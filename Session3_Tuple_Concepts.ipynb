#  1. Introduction to Tuples

A **tuple** is an ordered, immutable collection of elements.  
Tuples can store multiple items of different data types and are defined using **parentheses ( )**.

Once created **cannot be modified**.


```python
# Tuple creation examples

even_no_tuple=(2,4,6,8,10)
print(even_no_tuple)
```

```python
prg_tuple=("Excel","PowerBI","MYSQL","Python")
prg_tuple
```

```python
type(even_no_tuple)
```

```python
type(prg_tuple)
```

```python
#tuple with mixed datatype
mix_dtype_tuple=(23,"Python",34.3,False,78,"Ram","EXCEL")
mix_dtype_tuple
```

```python
type(mix_dtype_tuple)
```

```python
# Accessing tuple elements using indexing and slicing

   # First element
print(f'First item in mixed dtype tuple:{mix_dtype_tuple[0]}')

  # Last element
print(f'Last Item:{mix_dtype_tuple[-1]}')


  # Slice elements
print(f'Tuple item from 2nd position to 4th:{mix_dtype_tuple[2:5]}')

 # Reverse tuple
print(f'Reverse tuple:{mix_dtype_tuple[::-1]}')
```

```python
mix_dtype_tuple[-2::-1]
```

```python
# reverse string from -2 position
mix_dtype_tuple[-2::-1]
```

```python
# reverse string from -2 position to -5
mix_dtype_tuple[-2:-5:-1]
```

# 2. Tuples vs Lists (Immutable vs Mutable)

| Feature | Tuple | List |
|----------|--------|------|
| Syntax | `( )` | `[ ]` |
| Mutability | Immutable | Mutable |
| Performance | Faster | Slower (slightly) |
| Methods | Limited | Many |
| Use Case | Fixed data | Dynamic data |

Let’s see this difference in action.


```python
num_tuple = (2, 4, 5, 7, 5, 2, 6, 8)

# Convert to list
num_list = list(num_tuple)

# Modify the first element
num_list[0] = 10

# Convert back to tuple
num_tuple = tuple(num_list)
print(num_tuple)
```

# 3. Tuple Methods

Tuples have very few methods because they are immutable.


```python
# count() - counts the occurrence of a value


  # Count of '6'
print(num_tuple.count(6))

# index() - returns the first index of a value

   # Index of '5'
print(num_tuple.index(5))
```

# 💼 Final Case Study  
## 🏫 Student Management & Academic Record System

### 🎯 Objective:
Understand how to use **Tuples** (immutable data) and **Lists** (mutable, structured data) together  
to manage and organize information efficiently.

---

### 🧩 Scenario:

A school wants to maintain student data and marks for analysis.

- Each student’s personal details are **fixed** (Name, Roll Number, Course) → stored in a **Tuple**.
- Each student’s subject marks are **modifiable** → stored in a **List**.
- A collection of all students’ marks is stored as a **2D List**.
- For advanced organization, a **3D List** is used to represent marks for multiple batches.

---




```python
#Step 1: Create Tuple for each student (immutable personal info)
student1 = ("San", 101, "BCA")
student2 = ("Uma", 102, "BBA")
student3 = ("Vengi", 103, "BSc")

#Step 2: Create 1D List for marks of each student
stu_mark1=[88,98,100]
stu_mark2=[78,89,80]
stu_mark3=[68,90,90]


#Step 3: Combine all marks into a 2D List
batch1_marks=[stu_mark1,stu_mark2,stu_mark3]

print("Batch 1 Marks:", batch1_marks)


#Step 4: Combine all batches (3D List - two different batches)
batch2_marks = [
    [70, 75, 80],
    [88, 84, 82],
    [91, 93, 89]
]

print("Batch 2 Marks:", batch2_marks)



#Step 5: Display formatted data
all_batches = [batch1_marks, batch2_marks]


print("All Batches Mark")


print(all_batches)
```
